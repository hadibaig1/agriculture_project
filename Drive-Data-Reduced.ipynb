```python
import shap
import matplotlib.pyplot as plt

fig, axs = plt.subplots(1, 3, figsize=(20, 8))

for i in range(3):
    axs[i].set_title(f'Shap Values for class: {i}')
    shap.plots.bar(xgb_shap[2022][:, :, i], ax=axs[i], show=False)
plt.tight_layout()
plt.show()
```

```python
fig, axes  = plt.subplots(2, 2, figsize=(20,10))
axs = axes.flatten()

for i in range(3):
    axs[i].set_title(f'Shap Values for CLass: {i}')
    shap.plots.beeswarm(xgb_shap[2022][:, :, i], ax=axs[i], show=False, plot_size=None)

for j in range(3, len(axs.flat)):
    axs.flat[j].remove()
plt.tight_layout()
plt.show()
```

```python
# Now we are going to make these plots for the rest of the years as well
years = [2018, 2019, 2020, 2021, 2022]
class_dict = {
    0: 'Barley',
    1: 'Oats',
    2: 'Spring Wheat'
}
# Looping over all years
for year in years:
    fig, axes  = plt.subplots(2, 2, figsize=(20,10))
    axs = axes.flatten()

    for i in range(3):
        axs[i].set_title(f'{year} - Shap Values for Class: {class_dict[i]}')
        shap.plots.beeswarm(xgb_shap[year][:, :, i], ax=axs[i], show=False, plot_size=None)

    for j in range(3, len(axs.flat)):
        axs.flat[j].remove()
plt.tight_layout()
plt.show()
```

```python
# Finding only the common features among all years on average
from functools import reduce
dfs = [xgb_importances[2018], xgb_importances[2019], xgb_importances[2020], xgb_importances[2021], xgb_importances[2022]]

# Find common featurs to all 5 dfs
common_features = reduce(
    lambda acc, df: acc.intersection(set(df['feature'])),
    dfs[1:],
    set(dfs[0]['feature'])
)

# Filter each df to onlt keep tjose common features
filtered_dfs = [df[df['feature'].isin(common_features)] for df in dfs]
```

```python
# Accesing each individual df
df_2018_filtered, df_2019_filtered, df_2020_filtered, df_2021_filtered, df_2022_filtered = (filtered_dfs)
```
