#### Now we are going to extract the mean value of NDVI for each month for winter wheat for each year, and see why 2019 was predicted so badly

```python
# Now we are getting NDVI for each month to see how it changed
ndvis = ['0_ndwi', '1_ndwi', '2_ndwi', '3_ndwi', '4_ndvi', '5_ndvi']
years = {
    0: 'April',
    1: 'May',
    2: 'June',
    3: 'July',
    4: 'August',
    5: 'September'
}


all_ndvis = []
for month, ndvi in enumerate(ndvis):
    current_year = years[month]
    current_ndvi = winter_wheat_2019[ndvi]
    all_ndvis.append(f'{current_year}, NDVI: {current_ndvi.mean()}')
```

```python
all_ndvis
```

```python
all_ndvis_2018 = []
for month, ndvi in enumerate(ndvis):
    current_year = years[month]
    current_ndvi = winter_wheat_2018[ndvi]
    all_ndvis_2018.append(f'{current_year}, NDVI: {current_ndvi.mean()}')


all_ndvis_2020 = []
for month, ndvi in enumerate(ndvis):
    current_year = years[month]
    current_ndvi = winter_wheat_2020[ndvi]
    all_ndvis_2020.append(f'{current_year}, NDVI: {current_ndvi.mean()}')


all_ndvis_2021 = []
for month, ndvi in enumerate(ndvis):
    current_year = years[month]
    current_ndvi = winter_wheat_2021[ndvi]
    all_ndvis_2021.append(f'{current_year}, NDVI: {current_ndvi.mean()}')


all_ndvis_2022 = []
for month, ndvi in enumerate(ndvis):
    current_year = years[month]
    current_ndvi = winter_wheat_2022[ndvi]
    all_ndvis_2022.append(f'{current_year}, NDVI: {current_ndvi.mean()}')
```

```python
top_spring_2019 = spring_wheat_2019[top_10_features]
top_winter_2019 = winter_wheat_2019[top_10_features]
```

```python
# Now we will first scale our data with Standard sclaer so that we can plot the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scaled_top_winter_2019 = pd.DataFrame(scaler.fit_transform(top_winter_2019))
scaled_top_winter_2019.columns = top_10_features
pd.DataFrame(scaled_top_winter_2019)


scaled_top_spring_2019 = pd.DataFrame(scaler.transform(top_spring_2019))
scaled_top_spring_2019.columns = top_10_features
pd.DataFrame(scaled_top_spring_2019)
```

```python
# Group every 500 rows together and calculate the mean
smoothed_winter_2019 = scaled_top_winter_2019.groupby(scaled_top_winter_2019.index // 100).mean()
smoothed_spring_2019 = scaled_top_spring_2019.groupby(scaled_top_spring_2019.index // 100).mean()

# Now you can plot the smoothed data
smoothed_spring_2019.plot()
```
