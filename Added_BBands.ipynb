```python
# Finding only the common features among all years on average
from functools import reduce
dfs = [xgb_importances[2018], xgb_importances[2019], xgb_importances[2020], xgb_importances[2021], xgb_importances[2022]]

# Find common featurs to all 5 dfs
common_features = reduce(
    lambda acc, df: acc.intersection(set(df['feature'])),
    dfs[1:],
    set(dfs[0]['feature'])
)

# Filter each df to onlt keep tjose common features
filtered_dfs = [df[df['feature'].isin(common_features)] for df in dfs]
```
