```python
# Function to align a weather dataframe to the corresponding normal dataframe
def match_weather_to_normal(normal_df, weather_df):
    # Ensure consistent column set and order
    normal_cols = list(normal_df.columns)
    if 'class' not in weather_df.columns:
        weather_df['class'] = np.nan
    weather_df = weather_df.reindex(columns=normal_cols)

    # Match per-class counts from the normal dataframe
    normal_counts = normal_df['class'].value_counts().to_dict()
    parts = []
    for cls, needed in normal_counts.items():
        weather_cls = weather_df[weather_df['class'] == cls]
        have = len(weather_cls)
        if have == needed:
            parts.append(weather_cls)
        elif have > needed:
            parts.append(weather_cls.sample(n=needed, random_state=42))
        elif have > 0:
            parts.append(weather_cls.sample(n=needed, replace=True, random_state=42))
        else:
            # If weather has zero rows for this class, create fillers with NaNs except class
            filler = pd.DataFrame({c: [np.nan] * needed for c in normal_cols})
            filler['class'] = cls
            parts.append(filler)

    matched = pd.concat(parts, ignore_index=True)
    matched = matched[normal_cols]
    return matched

# Apply the matcher for each year and verify shapes and class counts
weather_df_2018_matched = match_weather_to_normal(df_2018, weather_df_2018)
weather_df_2019_matched = match_weather_to_normal(df_2019, weather_df_2019)
weather_df_2020_matched = match_weather_to_normal(df_2020, weather_df_2020)
weather_df_2021_matched = match_weather_to_normal(df_2021, weather_df_2021)
weather_df_2022_matched = match_weather_to_normal(df_2022, weather_df_2022)

for year, normal, weather_matched in [
    ('2018', df_2018, weather_df_2018_matched),
    ('2019', df_2019, weather_df_2019_matched),
    ('2020', df_2020, weather_df_2020_matched),
    ('2021', df_2021, weather_df_2021_matched),
    ('2022', df_2022, weather_df_2022_matched),
]:
    print(year, 'normal shape ->', normal.shape, 'weather matched shape ->', weather_matched.shape)
    print(' normal class counts:', normal['class'].value_counts().to_dict())
    print(' weather class counts:', weather_matched['class'].value_counts().to_dict())

# Recombine matched weather frames and compare to the combined normal DF
weather_df_matched = pd.concat([
    weather_df_2018_matched,
    weather_df_2019_matched,
    weather_df_2020_matched,
    weather_df_2021_matched,
    weather_df_2022_matched,
], ignore_index=True)
df_all = pd.concat([df_2018, df_2019, df_2020, df_2021, df_2022], ignore_index=True)
print('All normal shape:', df_all.shape, 'All weather matched shape:', weather_df_matched.shape)
```

```python
def match_weather_keep_weather(normal_df, weather_df):
    normal_counts = normal_df['class'].value_counts().to_dict()
    parts = []
    for cls, needed in normal_counts.items():
        wcls = weather_df[weather_df['class'] == cls]
        have = len(wcls)
        if have >= needed:
            parts.append(wcls.sample(n=needed, random_state=42))
        elif have > 0:
            parts.append(wcls.sample(n=needed, replace=True, random_state=42))
        else:
            filler = pd.DataFrame({c: [np.nan] * needed for c in weather_df.columns})
            filler['class'] = cls
            parts.append(filler)
    matched = pd.concat(parts, ignore_index=True)
    # ensure normal columns exist (as NaN) but keep weather-only columns
    for c in normal_df.columns:
        if c not in matched.columns:
            matched[c] = np.nan
    weather_only = [c for c in matched.columns if c not in normal_df.columns]
    matched = matched[list(normal_df.columns) + weather_only]
    return matched

# Example:
weather_df_2018_preserved = match_weather_keep_weather(df_2018, weather_df_2018)
```
